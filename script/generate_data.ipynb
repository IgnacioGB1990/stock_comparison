```python
import pandas as pd
import yfinance as yf
import os

# -----------------------------
# Parameters
# -----------------------------
START_DATE = "2020-07-14"
END_DATE = "2025-07-14"

# -----------------------------
# Dow Jones Tickers
# -----------------------------
tickers = [
    "AAPL", "AMGN", "AXP", "BA", "CAT", "CRM", "CSCO", "CVX",
    "DIS", "DOW", "GS", "HD", "HON", "IBM", "INTC", "JNJ",
    "JPM", "KO", "MCD", "MMM", "MRK", "MSFT", "NKE", "PG",
    "TRV", "UNH", "V", "VZ", "WBA", "WMT"
]

# -----------------------------
# Download data
# -----------------------------
all_data = []

for ticker in tickers:

    try:

        stock = yf.Ticker(ticker)

        # Company information
        info = stock.info
        company_name = info.get("longName", "")
        sector = info.get("sector", "")

        # Historical prices
        df = stock.history(
            start=START_DATE,
            end=END_DATE,
            auto_adjust=False,
            actions=True
        )

        if df.empty:
            print(f"No data for {ticker}")
            continue

        df = df.reset_index()

        df["Ticker"] = ticker
        df["Sector"] = sector
        df["CompanyName"] = company_name

        all_data.append(df)

        print(f"{ticker} downloaded")

    except Exception as e:
        print(f"Error downloading {ticker}: {e}")

# -----------------------------
# Combine all data
# -----------------------------
final_df = pd.concat(all_data, ignore_index=True)

# Select required columns
final_df = final_df[
    [
        "Date",
        "Open",
        "High",
        "Low",
        "Close",
        "Volume",
        "Dividends",
        "Stock Splits",
        "Ticker",
        "Sector",
        "CompanyName"
    ]
]

# -----------------------------
# Create output folder
# -----------------------------
output_folder = "data"
os.makedirs(output_folder, exist_ok=True)


# -----------------------------
# Export to CSV
# -----------------------------
output_file = os.path.join(output_folder, "dow_jones_data.csv")

final_df.to_csv(output_file, index=False, sep=";", decimal=",")

print(f"\nCSV successfully saved to: {output_file}")
print(f"Rows: {len(final_df):,}")
print(final_df.head())
```
